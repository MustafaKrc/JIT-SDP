```python
# remove rows from df where 'author_name' is null, None, empty string or in BOTS

df_removed = df[df['author_name'].isnull() | df['author_name'].isin(BOTS)]

df = df[~df['author_name'].isnull() & 
    ~df['author_name'].isin(BOTS) & 
    (df['author_name'] != '')]
```

```python
# get fileschanged column and split it into list of files changed by CAS_DELIMITER
# then remove empty strings from the list
# then remove preceeding and proceeding commas in each file name
CAS_DELIMITER = 'CAS_DELIMITER'
df['fileschanged'] = df['fileschanged'].apply(lambda x: [f.strip(',') for f in filter(None, x.split(CAS_DELIMITER))] if pd.notna(x) else [])
```

```python
df.iloc[4]['fileschanged']
```

```python
# for each file in df['fileschanged'] column, select rows only containing java files

df = df[df['fileschanged'].apply(lambda x: any([f.endswith('.java') for f in x]))]
```

```python

#df[df['fileschanged'].apply(lambda x: all(['test' in f.lower() for f in x]))]
df[df['fix'] == True]
```

```python
# remove rows where 'test' is in the file name of every file in 'fileschanged' column
df = df[df['fileschanged'].apply(lambda x: not all(['test' in f.lower() for f in x]))]

# print 5 rows that are filtered out by this rule
```

```python
# remove rows where 'example' is in the file name of every file in 'fileschanged' column
df = df[df['fileschanged'].apply(lambda x: not all(['example' in f.lower() for f in x]))]
```
